```python
import pylidc as pl
print(pl.__version__)
scans = pl.query(pl.Scan)
print(f"{scans.count()} scans were selected")
```

```python
import numpy as np
import matplotlib.pyplot as plt

from pylidc.utils import consensus
from skimage.measure import find_contours

for scan in scans:
    nods = scan.cluster_annotations()
    problematic_nods = [anns for anns in nods if len(anns) > 4]

    if not problematic_nods:
        continue

    vol = scan.to_volume(verbose=False)

    for anns in problematic_nods:
        print(scan.patient_id, len(anns))

        cmask, cbbox, masks = consensus(
            anns,
            clevel=0.5,
            pad=[(20, 20), (20, 20), (0, 0)],
        )

        colors = plt.cm.tab20(np.linspace(0, 1, len(masks)))

        for k in range(cmask.shape[2]):
            if not any(mask[:, :, k].any() for mask in masks):
                continue

            fig, ax = plt.subplots(figsize=(6, 6))
            ax.imshow(vol[cbbox][:, :, k], cmap="gray", alpha=0.7)

            for ann, mask, color in zip(anns, masks, colors):
                contours = find_contours(mask[:, :, k].astype(float), 0.5)

                for index, contour in enumerate(contours):
                    ax.plot(
                        contour[:, 1],
                        contour[:, 0],
                        color=color,
                        label=f"Annotation {ann.id}" if index == 0 else None,
                    )

            for index, contour in enumerate(
                find_contours(cmask[:, :, k].astype(float), 0.5)
            ):
                ax.plot(
                    contour[:, 1],
                    contour[:, 0],
                    "--k",
                    label="50% consensus" if index == 0 else None,
                )

            original_slice = cbbox[2].start + k
            ax.set_title(
                f"{scan.patient_id} — slice {original_slice} "
                f"— {len(anns)} annotations"
            )
            ax.axis("off")
            ax.legend()
            plt.show()
```
